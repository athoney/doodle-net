```python
n = [2, 3, 3, 1]
print(f"layer 1: {n[0]}")
print(f"layer 2: {n[1]}")
print(f"layer 3: {n[2]}")
print(f"layer 4: {n[3]}")
```

```python
W1 = np.random.randn(n[1], n[0])
W2 = np.random.randn(n[2], n[1])
W3 = np.random.randn(n[3], n[2])
b1 = np.random.randn(1, n[1])
b2 = np.random.randn(1, n[2])
b3 = np.random.randn(1, n[3])
# Standard shape = (1, n_units)
print("Weights for layer 1 shape:", W1.shape)
print("Weights for layer 2 shape:", W2.shape)
print("Weights for layer 3 shape:", W3.shape)
print("bias for layer 1 shape:", b1.shape)
print("bias for layer 2 shape:", b2.shape)
print("bias for layer 3 shape:", b3.shape)
```

```python
def sigmoid(arr):
    return 1 / (1 + np.exp(-1 * arr))
```

```python
# number of samples
m = 10

A0 = X
# layer 1
Z1 = A0 @ W1.T + b1
# (n_samples, n_nodes in layer 1)
assert Z1.shape == (m, n[1])
A1 = sigmoid(Z1)
print(Z1)
print(A1)

# layer 2
Z2 = A1 @ W2.T + b2
assert Z2.shape == (m, n[2])
A2 = sigmoid(Z2)

# layer 3
Z3 = A2 @ W3.T + b3
assert Z3.shape == (m, n[3])
A3 = sigmoid(Z3)

y_hat = A3
y_hat
```
